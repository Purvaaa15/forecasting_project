```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

pd.set_option("display.max_columns", None)
pd.set_option("display.width", 200)
```

```python
def assign_year(month, fy_start_year):
    """
    Financial year starts in April.
    Jan–Mar belong to next calendar year.
    """
    if month in ["January", "February", "March"]:
        return fy_start_year + 1
    else:
        return fy_start_year
```

```python
def prepare_monthly_series(file_path, fy_start_year):
    # Load Excel
    df = pd.read_excel(file_path)

    # Keep only required columns and rename
    df = df.rename(columns={
        "Content.Date": "month",
        "Content.Billed Quantity": "quantity"
    })

    # Month mapping
    month_map = {
        "January": 1, "February": 2, "March": 3, "April": 4,
        "May": 5, "June": 6, "July": 7, "August": 8,
        "September": 9, "October": 10, "November": 11, "December": 12
    }

    # Convert month name → number
    df["month_num"] = df["month"].map(month_map)

    # Assign correct calendar year using FY logic
    df["year"] = df["month"].apply(lambda m: assign_year(m, fy_start_year))

    # Create proper datetime index (1st day of month)
    df["date"] = pd.to_datetime(
        dict(year=df["year"], month=df["month_num"], day=1)
    )

    # Aggregate monthly demand
    monthly = (
        df.groupby("date", as_index=False)["quantity"]
        .sum()
        .sort_values("date")
    )

    return monthly
```

```python
monthly_2021 = prepare_monthly_series(r"D:\forecasting_project\data\raw\sales_2021.xlsx", 2021)  # FY 2021–22
monthly_2022 = prepare_monthly_series(r"D:\forecasting_project\data\raw\sales_2022.xlsx", 2022)  # FY 2022–23
monthly_2023 = prepare_monthly_series(r"D:\forecasting_project\data\raw\sales_2023.xlsx", 2023)  # FY 2023–24
monthly_2024 = prepare_monthly_series(r"D:\forecasting_project\data\raw\sales_2024.xlsx", 2024)  # FY 2024–25
```

```python
full_series = (
    pd.concat([monthly_2021, monthly_2022, monthly_2023, monthly_2024])
      .sort_values("date")
      .reset_index(drop=True)
)

full_series
```

XGBoost for Time Series Forecasting

```python
# Create a copy for ML
ml_df = full_series.copy()

# Create lag features
for lag in [1, 2, 3, 6, 12]:
    ml_df[f"lag_{lag}"] = ml_df["quantity"].shift(lag)

ml_df.head(15)
```

```python
ml_df = ml_df.dropna().reset_index(drop=True)

ml_df.shape
```

```python
X = ml_df.drop(columns=["date", "quantity"])
y = ml_df["quantity"]
```

```python
train_size = int(len(ml_df) * 0.8)

X_train, X_test = X.iloc[:train_size], X.iloc[train_size:]
y_train, y_test = y.iloc[:train_size], y.iloc[train_size:]

X_train.shape, X_test.shape
```

```python
from xgboost import XGBRegressor

xgb_model = XGBRegressor(
    n_estimators=300,
    max_depth=4,
    learning_rate=0.05,
    subsample=0.8,
    colsample_bytree=0.8,
    objective="reg:squarederror",
    random_state=42
)

xgb_model.fit(X_train, y_train)
```

```python
y_pred = xgb_model.predict(X_test)
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error
import numpy as np

mae = mean_absolute_error(y_test, y_pred)
rmse = np.sqrt(mean_squared_error(y_test, y_pred))
mape = (abs((y_test - y_pred) / y_test)).mean() * 100

mae, rmse, mape
```

```python
plt.figure(figsize=(12,5))
plt.plot(y_test.values, label="Actual", marker="o")
plt.plot(y_pred, label="XGBoost Prediction", marker="o")
plt.legend()
plt.grid(True)
plt.show()
```

```python
import pandas as pd

feature_importance = pd.Series(
    xgb_model.feature_importances_,
    index=X_train.columns
).sort_values(ascending=False)

feature_importance
```

```python
#MULTI-STEP FORECASTING
n_steps = 6

last_row = ml_df.iloc[-1:].copy()
future_predictions = []

for step in range(n_steps):
    # Prepare features
    X_future = last_row.drop(columns=["date", "quantity"])
    
    # Predict
    y_future = xgb_model.predict(X_future)[0]
    future_predictions.append(y_future)
    
    # Shift lag features safely
    last_row["lag_12"] = last_row["lag_6"]
    last_row["lag_6"]  = last_row["lag_3"]
    last_row["lag_3"]  = last_row["lag_2"]
    last_row["lag_2"]  = last_row["lag_1"]
    last_row["lag_1"]  = y_future
    
    # Update target
    last_row["quantity"] = y_future

future_predictions
```

```python
future_dates = pd.date_range(
    start=full_series["date"].iloc[-1],
    periods=n_steps + 1,
    freq="MS"
)[1:]
```

```python
plt.figure(figsize=(12,5))
plt.plot(full_series["date"], full_series["quantity"], label="Actual")
plt.plot(future_dates, future_predictions, marker="o", label="XGBoost Forecast")
plt.legend()
plt.grid(True)
plt.show()
```
